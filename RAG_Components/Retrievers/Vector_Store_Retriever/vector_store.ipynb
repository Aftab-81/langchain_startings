```python
from langchain_community.vectorstores import Chroma
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_core.documents import Document
```

```python
docs = [
    Document(
    page_content="Battle Through the Heavens Chinese: 斗破苍穹; pinyin: Dòupò Cāngqióng; lit. 'Fight Through The Sky' has main hero Xiao Zhan", 
    metadata= {'title': 'Battle Through the Heavens'}
    ),
    Document(page_content='Lin Yun (Chinese: 林允; pinyin: Lín Yǔn, born April 16, 1996), also known by her English name Jelly Li', metadata={'title': 'Lin Yun'}),
    Document(page_content='Su Qing (Chinese: 苏青, born 5 July 1989) is a Chinese actress known for her roles as Zhang Yan in Bea', metadata={'title': 'Su Qing (actress)'}),
    Document(page_content='This is a list of Chinese television series by genre. For a chronological list, see List of Chinese ', metadata={'title': 'List of Chinese television series'})
]
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embedding_funtion = HuggingFaceEmbeddings(model_name = "all-MiniLM-L6-v2")
```

```python
chroma_vector_store = Chroma.from_documents(
    documents = docs,
    embedding = embedding_funtion,
    collection_name = "my_collection"
)
```

```python
retriever = chroma_vector_store.as_retriever(search_kwargs = {"k": 2}, search_type = "mmr")
```

```python
query = "Xiao Zhan is my favourite character"
result = retriever.invoke(query)
result
```

```python
for i, doc in enumerate(result, start = 1):
    print(f"Document: {i}")
    print(doc)
```
