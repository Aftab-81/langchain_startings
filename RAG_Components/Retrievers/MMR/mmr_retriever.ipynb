```python
from langchain_community.vectorstores import FAISS
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_core.documents import Document
```

```python
# Sample documents
docs = [
    Document(page_content="Battle Through the Heavens follows Xiao Yan, a once-talented young cultivator who loses his power."),
    Document(page_content="Solo Leveling follows Sung Jin-Woo, the weakest hunter who gains the power to level up infinitely."),
    Document(page_content="Solo Leveling is set in a world where Hunters battle monsters that emerge from mysterious Gates."),
    Document(page_content="Classroom of the Elite follows Kiyotaka Ayanokoji, a seemingly ordinary student at an elite school with a hidden merit-based system."),
    Document(page_content="Classroom of the Elite explores manipulation, class rankings, and psychological strategy among students."),
    Document(page_content="Pokemon follows trainers who catch, train, and battle creatures called Pokemon across various regions."),
    Document(page_content="Pikachu, an Electric-type Pokemon, is one of the franchise's most iconic characters."),
]
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embedding_funtion = HuggingFaceEmbeddings(model_name = "all-MiniLM-L6-v2")
```

```python
vector_store = FAISS.from_documents(
    documents = docs,
    embedding = embedding_funtion
)
```

```python
retriever = vector_store.as_retriever(search_kwargs = {
    "k": 3,
    "lambda_mult": 1
}, search_type = "mmr")
```

```python
query = "Kiyotaka Ayanokoji is the masterpiece"
result = retriever.invoke(query)
result
```

```python
for i, doc in enumerate(result, 1):
    print(f"\n--- Result {i} ---")
    print(doc.page_content)
```
